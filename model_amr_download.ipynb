```python
from pathlib import Path

ROOT = Path("models/amr_abdi")

keywords = [
    "from_pretrained",
    "checkpoint",
    "model_path",
    "model_name",
    "huggingface",
    "drive.google",
    "gdown",
    "torch.load",
    ".bin",
    ".pt",
    ".pth"
]

for file in ROOT.rglob("*"):
    if file.is_file() and file.suffix in {".py", ".sh", ".md", ".yaml", ".yml", ".txt"}:
        try:
            text = file.read_text(encoding="utf-8")
            
            matches = []
            for i, line in enumerate(text.splitlines(), start=1):
                if any(k.lower() in line.lower() for k in keywords):
                    matches.append((i, line.strip()))
            
            if matches:
                print("\n" + "=" * 100)
                print(file)
                print("=" * 100)
                
                for i, line in matches:
                    print(f"{i:4d}: {line}")
                    
        except Exception:
            pass
```

```python
from pathlib import Path

ROOT = Path("models/amr_abdi")

keywords = [
    "TextToAMR(",
    "from_huggingface(",
    "repo_id",
    "huggingface.co/",
    "snapshot_download("
]

for file in ROOT.rglob("*"):
    if file.is_file() and file.suffix in {
        ".py", ".md", ".ipynb", ".txt", ".yaml", ".yml", ".sh"
    }:
        try:
            text = file.read_text(encoding="utf-8")

            matches = []

            for i, line in enumerate(text.splitlines(), start=1):
                if any(k.lower() in line.lower() for k in keywords):
                    matches.append((i, line.strip()))

            if matches:
                print("\n" + "=" * 100)
                print(file)
                print("=" * 100)

                for i, line in matches:
                    print(f"{i:4d}: {line}")

        except Exception:
            pass
```

## Load Model

```python
import sys
from pathlib import Path
import torch

# Path repository AMR
AMR_ROOT = Path("models/amr_abdi").resolve()

# Tambahkan repository AMR ke Python path
if str(AMR_ROOT) not in sys.path:
    sys.path.insert(0, str(AMR_ROOT))

# Import TextToAMR
from text_to_amr import TextToAMR

# Load model
text_to_amr = TextToAMR(
    model_name="mbart-en-id-smaller-fted",
    use_prefix=True
)

# Pastikan GPU tersedia
if not torch.cuda.is_available():
    raise RuntimeError("CUDA tidak tersedia.")

# Pindahkan model ke GPU
text_to_amr.model = text_to_amr.model.to("cuda:0")
text_to_amr.model.eval()

# Verifikasi
print("AMR Root :", AMR_ROOT)
print("Device   :", next(text_to_amr.model.parameters()).device)
print("GPU      :", torch.cuda.get_device_name(0))
```

```python
sentences = [
    "Presiden mengunjungi Jakarta."
]
graphs = text_to_amr(
    sentences,
    method=1
)
print(graphs[0])
```
